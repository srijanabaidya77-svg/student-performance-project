The main objectives of this stage are:

- Select relevant features for model development.
- Remove non-predictive identifier columns.

## Removing Non-Predictive Identifier Columns

Student_ID and Student_Name are identifiers rather than meaningful predictive features.

- Student_ID uniquely identifies each student.
- Student_Name identifies the student but does not provide useful information for predicting academic performance.

Therefore, these columns will be removed before preparing the dataset for machine learning.

```python
# Remove non-predictive identifier columns
df = df.drop(columns=["Student_ID", "Student_Name"])

print("Columns after removing identifiers:")
print(df.columns.tolist())

print("\nDataset Shape:", df.shape)
```

## Identifying Numerical and Categorical Features

Before encoding the dataset, the features are divided into two groups:

- **Numerical features:** Variables containing numerical values that can be used directly by machine learning models.
- **Categorical features:** Variables containing categories or labels that need to be converted into numerical form.

Identifying these groups helps us apply the appropriate preprocessing technique to each feature.

```python
# Identify numerical and categorical features

numerical_features = df.select_dtypes(include=["int64", "float64"]).columns.tolist()
categorical_features = df.select_dtypes(include=["object"]).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

## Defining the Target Variable

For the regression task, `Final_Exam_Marks` is selected as the target variable.

The model will learn from the student's academic and behavioral features and predict the student's final examination marks.

`Final_Exam_Marks` must not be included among the input features because it is the value we want the model to predict.

```python
# Define features and target

X = df.drop(columns=["Final_Exam_Marks"])
y = df["Final_Exam_Marks"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget variable:")
print(y.head())
```

## Removing Target-Leakage Features

`Pass_Fail` and `Performance_Category` describe the student's final performance.

Since the objective is to predict `Final_Exam_Marks`, these columns should not be used as input features. Including them could introduce target leakage and make the model appear more accurate than it actually is.

Therefore, both columns are removed from the feature set.

```python
# Remove target-leakage columns

X = X.drop(columns=["Pass_Fail", "Performance_Category"])

print("Features after removing target-leakage columns:")
print(X.columns.tolist())

print("\nFeature shape:", X.shape)
```

## Splitting the Dataset into Training and Testing Sets

The dataset is divided into training and testing sets before applying preprocessing.

- The training set is used to train the machine learning model.
- The testing set is used to evaluate the model on unseen data.
- An 80/20 split is used, as specified in the project requirements.
- `random_state=42` is used to make the split reproducible.

```python
# Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)

print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)
```

## Identifying Features for Preprocessing

The input features contain both numerical and categorical variables.

Numerical features can be used directly by machine learning models, while categorical features need to be converted into numerical values.

The categorical features are:

- Gender
- Department
- Internet_Access
- Study_Resources

These features will be encoded in the next preprocessing step.

```python
# Define numerical and categorical input features

categorical_features = [
    "Gender",
    "Department",
    "Internet_Access",
    "Study_Resources"
]

numerical_features = [
    "Age",
    "Semester",
    "Study_Hours_Per_Day",
    "Attendance_Percentage",
    "Previous_Semester_Marks",
    "Assignment_Average",
    "Internal_Test_Marks",
    "Practical_Marks",
    "Participation_Score",
    "Sleep_Hours",
    "Backlogs"
]

print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)

print("\nNumber of categorical features:", len(categorical_features))
print("Number of numerical features:", len(numerical_features))
```

## Creating the Preprocessing Pipeline

The dataset contains both numerical and categorical features, so different preprocessing techniques are required.

- Numerical features will be standardized using `StandardScaler`.
- Categorical features will be converted into numerical values using `OneHotEncoder`.
- `ColumnTransformer` will combine both preprocessing steps into a single pipeline.

The preprocessing object will be fitted only on the training data to avoid data leakage.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Create preprocessing pipeline

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

## Applying the Preprocessing Pipeline

The preprocessing pipeline is fitted using only the training data.

After fitting:

- The training data is transformed using the learned preprocessing parameters.
- The testing data is transformed using the same parameters.
- This prevents information from the testing set from influencing the preprocessing process.

```python
# Fit the preprocessor on training data and transform both datasets

X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Training data after preprocessing:", X_train_processed.shape)
print("Testing data after preprocessing:", X_test_processed.shape)
```

## Verifying the Processed Features

After preprocessing, all categorical variables have been converted into numerical columns and the numerical features have been standardized.

The processed training and testing datasets should therefore contain only numerical values.

```python
# Verify the processed data

print("Processed training data type:", X_train_processed.dtype)
print("Processed testing data type:", X_test_processed.dtype)

print("\nTraining data contains NaN:", np.isnan(X_train_processed).any())
print("Testing data contains NaN:", np.isnan(X_test_processed).any())
```
