# 2. Operators

Operators are symbols used to perform operations on values and variables.

Types of operators:
- Arithmetic operators
- Comparison operators
- Logical operators
- Assignment operators

2.1 Arithmetic Operators

```python
study_hours = 4
sleep_hours = 7

total_hours = study_hours + sleep_hours
difference = study_hours - sleep_hours
product = study_hours * sleep_hours
division = study_hours / sleep_hours

print("Addition:", total_hours)
print("Subtraction:", difference)
print("Multiplication:", product)
print("Division:", division)
```

```python
marks = 85

print(marks // 10)   # Floor division
print(marks % 10)    # Remainder
print(marks ** 2)    # Power
```

2.2 Comparison Operators

```python
marks = 72
passing_marks = 40

print(marks > passing_marks)
print(marks < passing_marks)
print(marks >= passing_marks)
print(marks <= passing_marks)
print(marks == passing_marks)
print(marks != passing_marks)
```

2.3 Logical Operators

```python
attendance = 82
marks = 67

print(attendance >= 75 and marks >= 40)
print(attendance >= 75 or marks >= 40)
print(not attendance >= 75)
```

```python
attendance = 82
marks = 67

eligible = attendance >= 75 and marks >= 40

print("Eligible:", eligible)
```

2.4 Assignment Operators

```python
marks = 50

marks += 10
print(marks)

marks -= 5
print(marks)

marks *= 2
print(marks)
```

# 4. Loops

Loops are used to execute a block of code repeatedly.

Python provides two commonly used loops:

- while loop

4.1 for loop

```python
for i in range(5):
    print(i)
```

4.2 range(start, stop)

4.3 Loop through student marks

```python
marks = [65, 72, 81, 59, 90]

for mark in marks:
    print("Student Mark:", mark)
```

4.4 Combine loop + condition

```python
marks = [65, 72, 81, 59, 90]

for mark in marks:

    if mark >= 90:
        category = "Excellent"
    elif mark >= 75:
        category = "Very Good"
    elif mark >= 60:
        category = "Good"
    elif mark >= 50:
        category = "Average"
    else:
        category = "Needs Improvement"

    print(mark, "->", category)
```

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

4.6 break

4.7 continue

continue skips the current iteration and moves to the next one.

```python
marks = [65, 72, 81, 59, 90]

for mark in marks:

    if mark < 60:
        continue

    print(mark)
```

# 11. Exception Handling

Exception handling allows a program to handle errors without immediately terminating.

Python commonly uses:
- try
- except
- else
- finally

11.1 Basic try-except

```python
try:
    number = 25
    print("Number:", number)

except ValueError:
    print("Please enter a valid number.")
```

11.2 Division error

```python
try:
    number1 = 10
    number2 = 0

    result = number1 / number2

    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

11.3 try-except-else

```python
marks = 75

try:
    marks = int(marks)
    print("Marks:", marks)

except ValueError:
    print("Invalid marks")
```

11.4 finally

```python
try:
    marks = 75
    print("Marks:", marks)

except ValueError:
    print("Invalid input.")

finally:
    print("Program execution completed.")
```

# 13. CSV Handling

CSV stands for Comma-Separated Values.

CSV files are commonly used to store tabular data.

Python provides the `csv` module for reading and writing CSV files.

```python
import csv

students = [
    ["Name", "Marks", "Attendance"],
    ["Srijana", 78, 82],
    ["Soumya", 85, 90],
    ["Ram", 68, 79],
    ["Shyam", 62, 81],
    ["Amit", 62, 75]
]

with open("student_test.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerows(students)

print("CSV file created successfully.")
```

```python
with open("student_test.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
students = [
    {"name": "Rahul", "marks": 78},
    {"name": "Priya", "marks": 35},
    {"name": "Amit", "marks": 91}
]

def check_result(marks):

    if marks >= 40:
        return "Pass"
    else:
        return "Fail"


for student in students:

    result = check_result(student["marks"])

    print(
        f"{student['name']} -> "
        f"Marks: {student['marks']} -> "
        f"{result}"
    )
```
