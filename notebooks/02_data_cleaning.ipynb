## Finding: Duplicate Records and Student IDs

The dataset initially contained 34 exact duplicate rows and 40 duplicated Student IDs.

## Missing-Value Treatment

```python
# Check missing values after data cleaning

missing_summary = pd.DataFrame({
    "Missing_Count": df_cleaned.isna().sum(),
    "Missing_Percentage": (
        df_cleaned.isna().sum() / len(df_cleaned) * 100
    ).round(2)
})

missing_summary = missing_summary[
    missing_summary["Missing_Count"] > 0
].sort_values(
    "Missing_Count",
    ascending=False
)

missing_summary
```

Step 1 — Fill numerical columns with median

```python
# Numerical columns for median imputation

numeric_imputation_columns = [
    "Age",
    "Study_Hours_Per_Day",
    "Attendance_Percentage",
    "Previous_Semester_Marks",
    "Assignment_Average",
    "Internal_Test_Marks",
    "Practical_Marks",
    "Participation_Score",
    "Sleep_Hours",
    "Backlogs"
]

for column in numeric_imputation_columns:
    median_value = df_cleaned[column].median()

    df_cleaned[column] = df_cleaned[column].fillna(median_value)

    print(
        f"{column}: missing values filled with median = {median_value}"
    )
```

Step 2 — Handle Study_Resources and Internet_Access

```python
# Fill missing categorical values with the mode

categorical_imputation_columns = [
    "Internet_Access",
    "Study_Resources"
]

for column in categorical_imputation_columns:
    mode_value = df_cleaned[column].mode()[0]

    df_cleaned[column] = df_cleaned[column].fillna(mode_value)

    print(
        f"{column}: missing values filled with mode = {mode_value}"
    )
```

## Standardizing Categorical Variables

Some categorical variables contain multiple representations of the same category.

For example, `Male`, `MALE`, `male`, and `M` represent the same gender category.

Similarly, Internet Access and Study Resources contain variations in capitalization, abbreviations, and spacing.

These values are standardized to consistent category names before machine learning preprocessing.

```python
# Standardize Gender

gender_mapping = {
    "M": "Male",
    "MALE": "Male",
    "Male": "Male",
    "Male ": "Male",
    "male": "Male",
    "F": "Female",
    "FEMALE": "Female",
    "Female": "Female",
    "Female ": "Female",
    "female": "Female",
    "Other": "Other"
}

df_cleaned["Gender"] = df_cleaned["Gender"].replace(gender_mapping)


# Standardize Internet Access

internet_mapping = {
    "Y": "Yes",
    "YES": "Yes",
    "Yes": "Yes",
    "yes": "Yes",
    "N": "No",
    "NO": "No",
    "No": "No",
    "no": "No"
}

df_cleaned["Internet_Access"] = df_cleaned["Internet_Access"].replace(
    internet_mapping
)


# Standardize Study Resources

study_resources_mapping = {
    "Both": "Both",
    "Both ": "Both",
    "both": "Both",

    "Notes": "Notes Only",
    "Notes Only": "Notes Only",
    "notes only": "Notes Only",

    "Online": "Online Courses",
    "Online Courses": "Online Courses",
    "online courses": "Online Courses",

    "Text Books": "Textbooks",
    "Textbooks": "Textbooks",
    "textbooks": "Textbooks"
}

df_cleaned["Study_Resources"] = df_cleaned["Study_Resources"].replace(
    study_resources_mapping
)

print("Categorical variables standardized successfully!")
```
